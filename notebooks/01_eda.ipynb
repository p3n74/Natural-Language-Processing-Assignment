# Exploratory analysis — Letterboxd sentiment

Run after `python -m src.load_data ...`. Adjust paths if your CSV column names differ.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

train = pd.read_csv(Path("../data/processed/train.csv"))
train["len"] = train["text"].str.len()
train["label"].value_counts().plot(kind="bar", title="Train class balance")
plt.show()
train["len"].hist(bins=30)
plt.title("Review character length")
plt.show()
```
